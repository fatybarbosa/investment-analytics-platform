## SQL

La base SQLite est reconstruite par le pipeline. Elle n'est pas versionnée. Les requêtes lisent :

1. le dernier passage qualité ;
2. les contrôles de ce passage dont le statut demande une relecture ;
3. le registre des jeux chargés ;
4. les douze dernières observations du SARON, dans l'unité publiée ;
5. la répartition sectorielle de la feuille de positions du fonds ;
6. la couverture de change du référentiel, par devise de cotation ;
7. les séances de cours qu'un contrôle écarte, par symbole ;
8. la feuille du fonds rapprochée du registre ODD ;
9. les positions du fonds sans ligne dans le référentiel ;
10. le poids de chaque secteur dans chaque livre, avec une fonction de fenêtre ;
11. les cinq contributeurs et les cinq détracteurs de chaque livre ;
12. un contrôle : les poids de chaque livre et de chaque mois font un ;
13. la richesse cumulée recalculée en SQL par une requête récursive, comparée à la richesse publiée ;
14. les mouvements du dernier rééquilibrage ;
15. un contrôle : la rotation publiée est la somme des mouvements ;
16. les dépassements de plafond, aux décisions et entre deux décisions.

La feuille du fonds n'est pas le poids des trois livres.

`python -m src.reporting.data_dictionary` exécute ces requêtes, puis réécrit le dictionnaire. Une requête en erreur arrête la commande et laisse le dictionnaire précédent en place. Le rafraîchissement complet appelle cette étape après le rapport.
