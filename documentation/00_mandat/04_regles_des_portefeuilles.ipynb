Ce carnet fixe les règles de construction, de suivi et de performance des trois portefeuilles. La construction est la phase 5, le suivi la phase 6, la performance la phase 7. Il appartient à la phase 0 et se lit avec l'introduction. Les termes sont définis à l'endroit où ils apparaissent.

Les documents du 16 octobre 2026 sont dans ce dossier : le profil du client (`01_profil_investissement.ipynb`), la politique d'investissement (`02_investment_policy_statement.ipynb`) et la méthodologie d'éligibilité ODD, version 1.1 (`03_methodologie_eligibilite_odd.ipynb`). La devise de référence est le franc suisse. Les trois portefeuilles partagent le même univers, la liste des titres, et la même série de rendements en francs. Ce rendement est un rendement total : la variation du cours ajusté des dividendes, converti en francs au cours de fin de mois. Le dividende est réinvesti par le cours ajusté. Le rendement est brut : il ne déduit ni frais ni impôt à la source. Ce choix remplace une première version qui retenait le cours de clôture sans dividende. Cette version comparait des livres sans dividende à un fonds qui publie un rendement total, ce qui n'est pas la même base.

La performance financière reste le premier objectif. L'alignement sur les Objectifs de développement durable (ODD) est une condition d'entrée du mandat. Il n'augmente pas un poids, ne remplace pas l'analyse financière et ne compense ni une liquidité insuffisante ni un risque excessif. Aucun objectif n'est prioritaire. Le nombre d'objectifs couverts ne change pas le poids. Le mandat ne présente pas la détention d'une action comme la réalisation de l'objectif : il n'y a pas d'impact direct. Il ne présente pas le placement comme la cause d'un résultat qui n'aurait pas existé sans lui : il n'y a pas d'additionnalité.

Le client accepte une exposition en devises étrangères. Le profil laisse la couverture de change ouverte. La couverture de change est un contrat qui annule l'effet du cours d'une monnaie. La politique d'investissement la tranche : elle n'est pas systématique. La conversion en francs, sans ce contrat, est donc conforme.

Le profil interdit la vente à découvert, soit la vente d'un titre que l'on ne détient pas ; le levier, soit l'emprunt destiné à investir au-delà du capital ; les fonds cotés conçus pour amplifier un indice ou pour évoluer en sens inverse ; le capital-investissement, part dans une société qui n'est pas en bourse, et les autres titres non cotés, titres sans cours public ; les produits structurés complexes, contrats dont le gain dépend d'une formule ; les armes controversées. La politique d'investissement nomme en outre la dette privée, prêt qui n'est pas un titre coté, et elle écarte les instruments difficiles à vendre ou dont le prix n'est pas fiable.


## Limites des portefeuilles

La politique d'investissement autorise les actions internationales cotées, les fonds actions cotés, et la combinaison des deux. La référence stratégique est le MSCI ACWI Sustainable Development Index (Net). Net signifie : dividendes après impôt. C'est l'indice, la liste de règles à laquelle le mandat se compare. On ne l'achète pas. Il n'a pas de cours propre. Le proxy investissable est le fonds iShares MSCI Global Sustainable Development Goals, fonds coté dont le sigle est ETF. C'est le fonds qui suit cet indice. Il a un prix, une feuille de positions et un rendement publié. La comparaison publiée utilise ce fonds : sa feuille fournit le secteur et le pays, et son rendement publié, converti en francs, sert de comparaison. Cette feuille ne contient que des actions. Les trois livres publiés sont donc des livres d'actions. Les plafonds ci-dessous sont ceux de la politique, sauf pour le secteur. La politique écrit 20 % par secteur. Sur la liste Eligible qui a un historique CHF complet, un livre entièrement investi ne peut pas respecter ce plafond : les secteurs ne peuvent pas contenir à eux tous la totalité du livre sous le plafond d'émetteur de 5 %. Le fichier `mandate_feasibility.csv` donne la capacité de chaque secteur. Les livres sont donc construits avec un plafond de travail de 40 %, qui n'a pas été approuvé par le client. Le plafond écrit n'est pas modifié : le suivi publie son dépassement dans `limit_watch.csv`, colonne `sector_over_ips_cap`, et les rapports le présentent comme un écart ouvert. Les plafonds d'émetteur, de zone et de devise sont ceux de la politique. Le poids est la part du portefeuille dans un titre.

Le plafond de 5 % s'applique à chaque ligne de cotation. La politique appelle cela un plafond par émetteur. Deux cotations de la même société sont deux lignes, et chacune est plafonnée à 5 %.

Les zones sont au nombre de quatre : Amériques, Europe, Asie-Pacifique, Moyen-Orient et Afrique. Elles viennent du pays inscrit sur la feuille de positions. Un pays absent de cette liste est rangé dans « Non classé », et ce groupe reste plafonné à 50 %. Le secteur est le métier indiqué sur cette feuille, dans la langue de cette feuille. Un secteur absent est aussi rangé dans « Non classé ». La devise de cotation est la monnaie du cours en bourse.

Le rendement qui sert à juger ces poids est le rendement total en francs, avec le dividende réinvesti. Le fonds iShares est jugé sur le rendement de sa valeur liquidative, converti en francs, net de ses frais. Les deux bases sont proches mais pas identiques : les livres sont bruts de frais et d'impôt à la source, le fonds est net de ses frais.

| Règle | Application |
|---|---|
| Émetteur | 5 % au plus, par ligne de cotation |
| Secteur | 20 % au plus dans la politique. 40 % au plus dans la construction : plafond de travail, non approuvé par le client |
| Zone | 50 % au plus |
| Devise de cotation | 40 % au plus |
| Cash opérationnel | Liquide à l'intérieur du portefeuille : 0 %, donc sous le plafond de 2 %. La réserve du client est hors mandat. |
| Sens | Long only : poids positifs ou nuls. Entièrement investi : les poids font 100 %. |
| Rééquilibrage | Recalcul des poids au début de janvier, avril, juillet et octobre. Le livre ne trade qu'à ces dates : entre deux dates, les poids dérivent avec le marché. |
| Échantillon | Les 60 mois déjà observés, sans le mois noté |
| Covariance | Tableau de la façon dont les titres varient ensemble. Ledoit-Wolf (2004) le stabilise vers la variance moyenne quand les titres sont nombreux et les mois peu nombreux. Le détail est dans la phase 5. |
| Lignes des livres optimisés | 25 à 40. La cardinalité est le nombre de lignes dont le poids est positif. |
| Rotation brute | Somme des changements de poids : 12,5 % au plus à chaque rééquilibrage, soit 50 % sur l'année |

L'équipondéré part de parts égales, puis les ramène sous les plafonds. Cette projection redistribue le reste pour rester à 100 %. Ce livre conserve toute la liste retenue. La bande de 25 à 40 lignes ne lui est pas imposée : la politique ne dit pas quels titres en sortir.

Cette bande est imposée à deux livres. La volatilité minimale cherche le livre qui a le moins varié sur les 60 mois connus. Le Sharpe maximal cherche le plus de rendement par unité de variation sur ces mêmes mois. La variance est l'ampleur des variations d'un titre avant la racine qui donne la volatilité. Le solveur cherche ces poids. S'il dépasse 40 lignes, les 40 plus gros poids sont gardés. S'il en a moins de 25, des titres sont ajoutés selon la préférence du livre : variance la plus basse pour la volatilité minimale, excédent de rendement moyen le plus haut pour le Sharpe maximal. Les poids sont ensuite reprojetés.

La comparaison à l'équipondéré, sur l'échantillon, se fait avant de ramener le nombre de lignes entre 25 et 40, et avant la limite de rotation. Si cette remise en conformité éloigne le livre du meilleur résultat obtenu sur l'échantillon, la limite de la politique prime.

Entre deux rééquilibrages, les poids dérivent avec le marché, et un poids peut passer un plafond sans qu'aucun ordre ne soit passé : c'est un dépassement passif. La politique ne dit pas s'il est toléré jusqu'au rééquilibrage suivant. Le suivi le compte à part. La rotation est mesurée contre le livre que le marché a laissé, pas contre la décision précédente. À la date de transaction, les plafonds de poids priment sur le budget de rotation : quand les deux ne tiennent pas ensemble, la rotation dépasse et le suivi le dit.
